```python
import networkx as nx
import matplotlib.pyplot as plt

# Criar um Grafo simples
G = nx.Graph()

#Adicionar vértice
G.add_nodes_from(['Audrey', 'Bill', 'Cathy', 'Dave', 'Ilana', 'Harry', 'Gayle', 'Frank', 'Emily', 'Jeff'])
```

```python
G.add_edges_from([
    ('Audrey','Bill'),
    ('Audrey','Ilana'),
    ('Audrey','Gayle'),
    ('Bill','Emily'),
    ('Bill','Jeff'),
    ('Jeff','Harry'),
    ('Harry','Ilana'),
    ('Ilana','Dave'),
    ('Dave','Frank'),
    ('Dave','Emily'),
    ('Emily','Frank'),
    ('Jeff','Emily'),
    ('Cathy','Emily'),
    ('Cathy', 'Gayle')
])
```

```python
print(f"Nós do grafo: {G.nodes()}")
print(f"Aresta do grafo: {G.edges()}")
print(f"Contando o num de Nós: {G.number_of_nodes()}")
print(f"Contando o num de Arestas: {G.number_of_edges()}")
```

```python
pos = nx.spring_layout(G, seed=42)
nx.draw(G, pos, with_labels = True, node_color="lightgreen", font_weight="bold", node_size=700, font_size=10)

plt.title('Atividade 1')
plt.show()
```
