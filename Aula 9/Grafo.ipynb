```python
import networkx as nx
import matplotlib.pyplot as plt

# Criar um Grafo simples
G = nx.DiGraph()

#Adicionar vértice
G.add_nodes_from(['Prince', 'David Bowie', 'Madonna', 'Michael jackson', 'Beyonce', 'Kanye West', 'Jay-z', 'Rihanna', 'Lady Gaga'])
```

```python
G.add_edges_from([
    ('Prince','Michael jackson'),
    ('David Bowie','Prince'),
    ('Michael jackson','Beyonce'),
    ('Jay-z','Rihanna'),
    ('Kanye West','Rihanna'),
    ('Madonna','Lady Gaga'),
    ('David Bowie','Madonna'),
    ('Beyonce', 'Rihanna')

])
```

```python
print(f"Nós do grafo: {G.nodes()}")
print(f"Aresta do grafo: {G.edges()}")
print(f"Contando o num de Nós: {G.number_of_nodes()}")
print(f"Contando o num de Arestas: {G.number_of_edges()}")
```

```python
pos = nx.spring_layout(G, seed=42)
nx.draw(G, pos, with_labels = True, node_color="lightgreen", font_weight="bold", node_size=800, font_size=10, arrowsize=15)

plt.title('Atividade 1')
plt.show()
```
